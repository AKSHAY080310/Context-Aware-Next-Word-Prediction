```python
import nltk
```

```python
nltk.download('punkt_tab')
```

```python
from nltk.tokenize import word_tokenize
```

```python
sentence = "I Love you I Hate you"
```

```python
tokens=word_tokenize(sentence)
print(tokens)
```

```python
with open("/content/train_cleaned1.txt","r",encoding="utf-8") as f:
  train_clean=f.read().splitlines()
```

```python
for i in range(3):
  print(word_tokenize(train_clean[i]))
```

```python
train_tokens = []

for sentence in train_clean:
    train_tokens.append(word_tokenize(sentence))
```

```python
with open("/content/valid_cleaned1.txt","r",encoding="utf-8") as f:
  valid_clean=f.read().splitlines()
```

```python
with open("/content/test_cleaned1.txt","r",encoding="utf-8") as f:
  test_clean=f.read().splitlines()
```

```python
valid_tokens=[]
for sentence in valid_clean:
  valid_tokens.append(word_tokenize(sentence))
```

```python
test_tokens=[]
for sentence in test_clean:
  test_tokens.append(word_tokenize(sentence))
```

```python
for i in range(5):
    print(f"Sentence {i}")
    print(train_tokens[i])
    print("-" * 60)
```

```python
import pickle

with open("train_tokens.pkl", "wb") as f:
    pickle.dump(train_tokens, f)

with open("valid_tokens.pkl", "wb") as f:
    pickle.dump(valid_tokens, f)

with open("test_tokens.pkl", "wb") as f:
    pickle.dump(test_tokens, f)
```
