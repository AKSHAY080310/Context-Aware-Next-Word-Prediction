```python
import re
def clean_text(split):
  cleaned_text=[]
  for row in split:
    line=row["text"].strip()
    if line=="":
      continue
    if re.fullmatch(r"=+\s.*?\s=+" , line):
      continue
    line=re.sub(r"\s+", " ", line)
    cleaned_text.append(line)
  return cleaned_text
```

```python
import re
def clean_text(split):
    cleaned_text = []
    for row in split:
        line = row["text"].strip()
        if line == "":
            continue
        if re.fullmatch(r"=+\s.*?\s=+", line):
            continue
        line = line.replace(" @-@ ", "-")
        line = line.replace(" @,@ ", ", ")
        line = line.replace(" @.@ ", ".")
        line = re.sub(r"\s+", " ", line)
        cleaned_text.append(line)

    return cleaned_text
```

```python
train_cleaned = clean_text(train_data)
```

```python
valid_cleaned=clean_text(valid_data)
```

```python
test_cleaned=clean_text(test_data)
```

```python
def clean_text1(split):
  clean_text=[]
  for row in split:
    line=row["text"].strip()
    if line=="":
      continue
    if re.fullmatch(r"=+\s*.*?\s*=+",line):
      continue
    line=re.sub(r"\s*@-@\s*","-",line)
    line=re.sub(r"\s*@.@\s*",".",line)
    line=re.sub(r"(\d)\s*@,@\s*(\d)",r"\1,\2",line)
    line=re.sub(r"\s*@,@\s*",", ",line)
    line=re.sub(r"\s*@,@$",",",line)
    line=re.sub(r"\s+"," ",line).strip()
    clean_text.append(line)
  return clean_text
```

```python
train_cleaned1=clean_text1(train_data)
valid_cleaned1=clean_text1(valid_data)
test_cleaned1=clean_text1(test_data)
```

```python
for token in ["@-@","@.@","@,@"]:
  count=sum(token in sentence for sentence in train_cleaned1)
  print(f"{token}:{count}")
```

```python
def save_cleaned_data(data, filename):
    with open(filename, "w", encoding="utf-8") as f:
        for line in data:
            f.write(line + "\n")
```

```python
save_cleaned_data(train_clean, "train_clean.txt")
save_cleaned_data(valid_clean, "valid_clean.txt")
save_cleaned_data(test_clean, "test_clean.txt")
```
