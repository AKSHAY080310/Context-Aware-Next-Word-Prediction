```python
import pickle
with open("/content/train_tokens.pkl","rb") as f:
  train_tokens=pickle.load(f)
with open("/content/valid_tokens.pkl","rb") as f:
  valid_tokens=pickle.load(f)
with open("/content/test_tokens.pkl","rb") as f:
  test_tokens=pickle.load(f)
```

```python
print(train_tokens)
```

```python
print("Train tokens:", len(train_tokens))
print("Validation tokens:", len(valid_tokens))
print("Test tokens:", len(test_tokens))
```

```python
train_words = [word for sentence in train_tokens for word in sentence]
print(len(train_words))
```

```python
from collections import Counter
counter = Counter(train_words)
print(len(counter))
PAD_TOKEN="<PAD>"
UNK_TOKEN="<UNK>"
word_to_id = {
    PAD_TOKEN: 0,
    UNK_TOKEN: 1
}
for word in counter:
  if word not in word_to_id:
    word_to_id[word] = len(word_to_id)

id_to_word = {idx:word for word, idx in word_to_id.items()}
print("Vocabulary size:", len(word_to_id))
```

```python
def encode_tokens(tokenized_data, word_to_id):
  encoded_data = []

  for sentence in tokenized_data:
    encoded_sentence = [word_to_id.get(word, word_to_id["<UNK>"])
                        for word in sentence]
    encoded_data.append(encoded_sentence)
  return encoded_data

train_ids = encode_tokens(train_tokens, word_to_id)
valid_ids = encode_tokens(valid_tokens, word_to_id)
test_ids = encode_tokens(test_tokens, word_to_id)
```

```python
print("Original:")
print(train_tokens[20][:20])

print("\nEncoded:")
print(train_ids[20][:20])

print("\nDecoded:")
print([id_to_word[idx] for idx in train_ids[20][:20]])
```

```python
valid_unknowns = sum(
    1 for sentence in valid_tokens
    for word in sentence
    if word not in word_to_id
)
test_unknowns = sum(
    1 for sentence in test_tokens
    for word in sentence
    if word not in word_to_id

)
print("Validation unknowns:", valid_unknowns)
print("test unknowns:", test_unknowns)
```

```python
valid_unk_percentage = (valid_unknowns / len(valid_tokens)) * 100
test_unk_percentage = (test_unknowns / len(test_tokens)) * 100

print(f"Validation <UNK>: {valid_unk_percentage:.2f}%")
print(f"Test <UNK>: {test_unk_percentage:.2f}%")
```

```python
valid_total = sum(len(sentence) for sentence in valid_tokens)
test_total = sum(len(sentence) for sentence in test_tokens)

valid_unk_percentage = (valid_unknowns / valid_total) * 100
test_unk_percentage = (test_unknowns / test_total) * 100
print(len(valid_tokens))
print("Validation tokens:", valid_total)
print("Validation unknowns:", valid_unknowns)
print(f"Validation <UNK>: {valid_unk_percentage:.2f}%")

print()
print(len(test_tokens))
print("Test tokens:", test_total)
print("Test unknowns:", test_unknowns)
print(f"Test <UNK>: {test_unk_percentage:.2f}%")
```

```python
def create_context_windows(encoded_data, context_size):
    X = []
    y = []

    for sequence in encoded_data:

        # Skip sequences shorter than the context size + target
        if len(sequence) <= context_size:
            continue

        for i in range(len(sequence) - context_size):
            context = sequence[i:i + context_size]
            target = sequence[i + context_size]

            X.append(context)
            y.append(target)

    return X, y
```

```python
context_size = 10

X_train, y_train = create_context_windows(
    train_ids,
    context_size
)

X_valid, y_valid = create_context_windows(
    valid_ids,
    context_size
)

X_test, y_test = create_context_windows(
    test_ids,
    context_size
)
```
